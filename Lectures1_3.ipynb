```python
# For Loop
for rating in df["Customer_Rating"]:

    if rating >= 4:
        print("Good Rating")
    else:
        print("Needs Improvement")
```

```python
# FOR Loop with range() iloc= integer location

for i in range(20):
    print(df["Product"].iloc[i], df["Revenue_Lakh"].iloc[i])
```

```python
#WHILE Loop in Python
#While loop and i,j is variable to track
i = 0

while i < 5:
    revenue = df["Revenue_Lakh"].iloc[i]

    if revenue >= 300:
        print("High Revenue:", revenue)
    else:
        print("Low Revenue:", revenue)

    i = i + 1
```

```python
#While loop and i,j is variable to track
i = 0

while i < 5:
    j = 0

    while j < 4:
        print("Product:", df["Product"].iloc[i],
              "Quarter:", df["Quarter"].iloc[j])
        j = j + 1

    i = i + 1
```

```python
#i = outer counter j = inner counter

i = 0

while i < 3:
    print("Checking Product:", df["Product"].iloc[i])

    j = 1

    while j <= 2:
        print("Performance Check:", j)
        j = j + 1

    i = i + 1
```

```python
# Nested While looop
i = 0

while i < 5:              # Outer while loop
    print("Product:", df["Product"].iloc[i])

    j = 0

    while j < 2:          # Inner while loop
        if j == 0:
            print("Revenue:", df["Revenue_Lakh"].iloc[i])
        else:
            print("Customer Rating:", df["Customer_Rating"].iloc[i])

        j = j + 1

    i = i + 1
```

```python
# . COUNT PROFITABLE RECORDS
# Business Question: How many observations are profitable?
# ============================================================

count = 0

for profit in df["Profit_Lakh"]:

    if profit > 0:
        count = count + 1

print("Profitable Records:", count)
```

```python
# PROFIT / LOSS USING IF
# Business Question: Classify the first 10 records as
# Profit or Loss.
# ============================================================

for i in range(10):

    profit = df["Profit_Lakh"].iloc[i]

    if profit > 0:
        print(
            df["Product"].iloc[i],
            "→ Profit"
        )
    else:
        print(
            df["Product"].iloc[i],
            "→ Loss"
        )
```

```python
# ============================================================
#  CREATE PROFIT
# Business Question: Is the business making a profit or loss?
# ============================================================

df["Profit_Lakh"] = (
    df["Revenue_Lakh"] - df["Total_Cost_Lakh"]
)

print(
    df[
        ["Product", "Revenue_Lakh",
         "Total_Cost_Lakh", "Profit_Lakh"]
    ].head(10)
)
```

```python
# ============================================================
# GROUPBY() — STUDENTS
# Business Question: Which product has the highest average
# student enrollment?
# ============================================================

print(
    df.groupby("Product")["Students_Enrolled"].mean()
)
```
