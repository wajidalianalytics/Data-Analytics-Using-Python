# Section 1: Create Different Data Structures in Python

A **data structure** is a way of storing data so that we can use it easily.
Python has 4 built-in structures:

| Structure | Symbol | Ordered? | Changeable? | Duplicates? | Starbucks use |
|---|---|---|---|---|---|
| **List** | `[ ]` | Yes | Yes | Allowed | Menu items that can change |
| **Tuple** | `( )` | Yes | No | Allowed | A store's fixed details |
| **Set** | `{ }` | No | Yes | Not allowed | Unique cities |
| **Dictionary** | `{k: v}` | Yes | Yes | Keys unique | Product → Price |

### 📖 Definitions
| Term | Definition | Starbucks example |
|---|---|---|
| **Data structure** | A way of organising and storing data in memory so it can be used efficiently. | Storing menu, prices, cities |
| **List** | An *ordered* and *changeable* collection of items in square brackets `[ ]`. Duplicates allowed. | `["Latte", "Cold Brew"]` |
| **Tuple** | An *ordered* but *unchangeable* collection in round brackets `( )`. | `("S01", "Connaught Place", "Delhi")` |
| **Set** | An *unordered* collection of **unique** items in curly brackets `{ }`. | `{"Delhi", "Mumbai"}` |
| **Dictionary** | A collection of **key : value** pairs. Each key is unique and is used to find its value. | `{"Latte": 270}` |
| **Mutable** | Can be changed after creation (list, set, dict). | Add a new drink to the menu |
| **Immutable** | Cannot be changed after creation (tuple, string, number). | A store's fixed record |
| **Record** | One complete row of data, often stored as a dictionary. | One order with all its details |

````python
# ---- Part A: create them by typing ----

# LIST: the Starbucks menu (8 products)
menu = ["Cappuccino", "Latte", "Americano", "Cold Brew",
        "Frappuccino", "Masala Chai", "Blueberry Muffin", "Veg Sandwich"]

# TUPLE: one store's fixed details (StoreID, name, city)
store = ("S01", "Connaught Place", "Delhi")

# SET: cities where Starbucks sells (the duplicate 'Delhi' is removed automatically)
cities = {"Delhi", "Mumbai", "Bengaluru", "Hyderabad", "Pune", "Delhi"}

# DICTIONARY: product -> base price in Rs
price = {"Cappuccino": 250, "Latte": 270, "Americano": 220, "Cold Brew": 300,
         "Frappuccino": 350, "Masala Chai": 200, "Blueberry Muffin": 180, "Veg Sandwich": 240}

print("List      :", menu)
print("Tuple     :", store)
print("Set       :", cities, " <- 'Delhi' typed twice, stored once")
print("Dictionary:", price)
print()
print(type(menu), type(store), type(cities), type(price))
````

````python
# ---- Part B: create them from the Excel data ----
# Each Excel column becomes a Python list. Each Excel row becomes a dictionary.

product_list = df["Product"].tolist()        # 100 products (with repeats)
city_list    = df["City"].tolist()           # 100 cities  (with repeats)
qty_list     = df["Quantity"].tolist()       # 100 quantities
revenue_list = df["Revenue"].tolist()        # 100 revenue values

first_order   = df.iloc[0].to_dict()                 # ONE order  -> dictionary
first_5_orders = df.head(5).to_dict("records")       # FIVE orders -> list of dictionaries

print("Items in product_list :", len(product_list))
print("Unique cities (set)   :", set(city_list))
print()
print("First order as a dictionary:")
print(first_order)
````

# Section 2: Access Elements of Data Structures (6 min)

- **List / Tuple**: use the **position (index)**. Python starts counting from **0**.
- **Dictionary**: use the **key**.
- **Set**: has no position, so we check *membership* (`in`) or loop over it.

```
Index:    0           1        2          3            ...   7
menu = ["Cappuccino","Latte","Americano","Cold Brew", ... ,"Veg Sandwich"]
Negative: -8          -7       -6         -5           ...   -1
```

### 📖 Definitions
| Term | Definition | Starbucks example |
|---|---|---|
| **Element / Item** | One single value inside a data structure. | `"Latte"` in the menu |
| **Index** | The position number of an element. Starts from **0**. Negative index counts from the end (-1 = last). | `menu[0]` = Cappuccino |
| **Slicing** | Taking a part of a sequence using `[start:stop:step]`. The stop position is *not* included. | `menu[0:3]` = first three items |
| **Key** | The unique label used to find a value in a dictionary. | `price["Latte"]` |
| **Nested structure** | A structure inside another structure, e.g. a list of dictionaries. | `first_5_orders[2]["City"]` |
| **Membership test** | Checking if a value exists using `in`. Returns `True` or `False`. | `"Pune" in cities` |
| **`.get()`** | Dictionary method that returns a default value instead of an error when the key is missing. | `price.get("Pizza", "Not on menu")` |

````python
# --- List ---
print("First item      :", menu[0])
print("Last item       :", menu[-1])
print("First three     :", menu[0:3])      # slicing: start included, end excluded
print("Every 2nd item  :", menu[::2])

# --- Tuple (same as list) ---
print("\nStore ID        :", store[0])
print("Store city      :", store[2])

# --- Dictionary ---
print("\nPrice of Latte  :", price["Latte"])
print("Safe lookup     :", price.get("Pizza", "Not on menu"))   # no error if key is missing

# --- Set ---
print("\nIs Pune a Starbucks city?   ", "Pune" in cities)
print("Is Jaipur a Starbucks city? ", "Jaipur" in cities)
````

````python
# --- Changing data: a list is changeable, a tuple is not ---
menu[6] = "Muffin"
print("Updated menu :", menu)
menu[6] = "Blueberry Muffin"          # change it back, we need the original name later

try:
    store[1] = "Saket Mall"
except TypeError as e:
    print("Tuple error  :", e)
````

````python
# --- Access the real Excel data stored in Python structures ---
print("City of order #1       :", city_list[0])
print("Quantity of last order :", qty_list[-1])
print("First 5 revenues       :", revenue_list[:5])

print("\nProduct of first order :", first_order["Product"])
print("Revenue of first order :", first_order["Revenue"])

# list of dictionaries: [position][key]
print("\nCity of 3rd order      :", first_5_orders[2]["City"])
print("Product of 3rd order   :", first_5_orders[2]["Product"])
````

# Section 3: Apply Functions to Data Structures (7 min)

Python gives us ready-made **functions** (`len`, `sum`, `max`, `min`, `sorted`) and **methods** (`.append()`, `.keys()` ...) so we don't write everything ourselves.

### 📖 Definitions
| Term | Definition | Starbucks example |
|---|---|---|
| **Function** | A reusable block of code that takes input and gives output. Called as `name(data)`. | `sum(revenue_list)` |
| **Method** | A function that belongs to an object. Called as `object.name()`. | `menu.append("Iced Tea")` |
| **`len()`** | Counts the number of elements. | Number of orders |
| **`sum()` / `max()` / `min()`** | Total / largest / smallest value. | Total revenue, biggest order |
| **`sorted()`** | Returns a new list arranged in order. | Top 5 orders |
| **Loop (`for`)** | Repeats an action for every element. | Count orders in each city |
| **`zip()`** | Pairs elements of two sequences together. | Product with its revenue |
| **List comprehension** | A short one-line way to build a new list from an existing one. | `[r for r in revenue_list if r > 800]` |
| **Union `\|`** | All items from both sets. | Cities selling Cold Brew or Chai |
| **Intersection `&`** | Items present in **both** sets. | Cities selling both products |
| **Difference `-`** | Items in the first set but not in the second. | Cities with Cold Brew but no Chai |

````python
# Built-in functions on the Revenue list (100 orders)
print("Number of orders   :", len(revenue_list))
print("Total revenue      : Rs", round(sum(revenue_list), 2))
print("Average order value: Rs", round(sum(revenue_list) / len(revenue_list), 2))
print("Biggest order      : Rs", max(revenue_list))
print("Smallest order     : Rs", min(revenue_list))
print("Top 5 orders       :", sorted(revenue_list, reverse=True)[:5])
print("Total cups/items   :", sum(qty_list))
````

````python
# Count orders in each city

orders_per_city = {}

for c in city_list:

    if c in orders_per_city:
        orders_per_city[c] = orders_per_city[c] + 1
    else:
        orders_per_city[c] = 1

print("Orders per city:", orders_per_city)


# Find the busiest city
busiest_city = ""
highest_orders = 0

for c, orders in orders_per_city.items():

    if orders > highest_orders:
        highest_orders = orders
        busiest_city = c

print("Busiest city:", busiest_city)
````

````python
# Total revenue for each product
revenue_by_product = {}

for p, r in zip(product_list, revenue_list):

    if p in revenue_by_product:
        revenue_by_product[p] = revenue_by_product[p] + r
    else:
        revenue_by_product[p] = r

for p, r in revenue_by_product.items():
    print(p, "-> Rs", r)

# Find the product with the highest revenue
best_product = ""
highest_revenue = 0

for p, r in revenue_by_product.items():
    if r > highest_revenue:
        highest_revenue = r
        best_product = p

print("Best-selling product by revenue:", best_product)
````

````python
# Find orders above Rs 800
high_value = []

for r in revenue_list:
    if r > 800:
        high_value.append(r)

print("Orders above Rs 800:", len(high_value), "->", high_value)


# Find products priced above Rs 260
premium = []

for p in price:
    if price[p] > 260:
        premium.append(p)

print("Premium products:", premium)


# Add a product to the menu
menu.append("Iced Tea")

# Remove the product from the menu
menu.remove("Iced Tea")

# Count the number of products in the menu
print("\nMenu size:", len(menu))

# Show the first 3 product names from the dictionary
print("Keys:", list(price.keys())[:3])

# Show the first 3 prices from the dictionary
print("Values:", list(price.values())[:3])
````

````python
# Find cities where Cold Brew was sold
cold_brew_cities = set()

for c, p in zip(city_list, product_list):
    if p == "Cold Brew":
        cold_brew_cities.add(c)

# Find cities where Masala Chai was sold
chai_cities = set()

for c, p in zip(city_list, product_list):
    if p == "Masala Chai":
        chai_cities.add(c)

print("Cold Brew sold in:", cold_brew_cities)
print("Masala Chai sold in:", chai_cities)

# Cities where both products were sold
print("Both products:", cold_brew_cities & chai_cities)

# Cities where either product was sold
print("Any of the two:", cold_brew_cities | chai_cities)

# Cities where Cold Brew was sold but Masala Chai was not
print("Cold Brew but no Chai:", cold_brew_cities - chai_cities)
````

# Section 4: Evaluate the Utility of Data Structures (6 min)

**Which one should a Starbucks analyst choose?**

| Business situation (in our dataset) | Best choice | Why |
|---|---|---|
| Menu that changes often | **List** | Easy to add, remove, change |
| A store's fixed record (ID, name, city) | **Tuple** | Cannot be changed by mistake |
| Unique cities / payment modes | **Set** | No duplicates, very fast search |
| Product → price, or OrderID → full order | **Dictionary** | Instant lookup by name/ID |
| 100,000 prices for fast maths | **NumPy array** | Very fast calculations |
| The whole 100 × 15 sales table | **Pandas DataFrame** | Best for business analysis |

### 📖 Definitions
| Term | Definition | Starbucks example |
|---|---|---|
| **Utility** | How useful and suitable a data structure is for a given task. | Using a set to find unique payment modes |
| **Lookup / Search** | Finding whether a value exists or fetching its value. | Find order SBX1050 |
| **Duplicate** | The same value appearing more than once. | 'Delhi' appears many times in `city_list` |
| **Time efficiency** | How fast an operation runs. Sets and dictionaries search much faster than lists for large data. | Searching 1 million IDs |
| **Trade-off** | Every structure is good at something and weak at something else. Choose based on the task. | List keeps order, set removes duplicates |

# Section 5: NumPy Arrays (8 min)

**NumPy** stores numbers in an **array**: like a list, but much faster, and it can do maths on **all values at once** (no loop needed).

We now move the numeric columns of our Excel data (`Unit_Price`, `Quantity`, `Discount_Pct`, `Revenue`) into NumPy arrays.

### 📖 Definitions
| Term | Definition | Starbucks example |
|---|---|---|
| **NumPy** | *Numerical Python*: a library for fast mathematical operations on numbers. | `import numpy as np` |
| **Array (`ndarray`)** | A grid of values of the **same data type**, stored compactly for speed. | `unit_price` array |
| **1D / 2D array** | 1D = one row of values (like a list). 2D = rows and columns (like a table). | 2D: orders × 4 numeric columns |
| **Shape** | The size of the array as (rows, columns). | `(100, 4)` |
| **`ndim`** | Number of dimensions of the array. | 2 |
| **`dtype`** | The data type of the values inside the array. | `int64`, `float64` |
| **Vectorisation** | Doing maths on the whole array at once without a loop. | `unit_price * qty * (1 - discount)` |
| **Axis** | Direction of calculation. `axis=0` goes down the rows (per column). `axis=1` goes across (per row). | Average of each column |
| **Aggregation** | Reducing many values to one summary value. | `sum`, `mean`, `max` |
| **Boolean mask** | An array of True/False created from a condition, used to filter or count. | `qty >= 3` |
| **`NaN`** | "Not a Number": a missing value. `np.nanmean` ignores it. | 3 blank Ratings |

````python
# 1D arrays: one array per numeric column.
order_ids  = df["OrderID"].to_numpy()
unit_price = df["Unit_Price"].to_numpy()
qty        = df["Quantity"].to_numpy()
discount   = df["Discount_Pct"].to_numpy()

print("Shape :", unit_price.shape, "-> 100 values in one dimension")
print("Type  :", unit_price.dtype)
print("First 5 prices   :", unit_price[:5])
print("First 5 quantity :", qty[:5])
print("First 5 discount :", discount[:5])
````

````python
# Maths on the whole array at once (no loop). Rebuild the Revenue formula from Excel:it calculates the revenue for every order using NumPy, and second, it calculates some summary statistics such as total, average, median, and maximum revenue.
#it calculates the revenue for every order using NumPy,
#and second, it calculates some summary statistics such as total, average, median, and maximum revenue.

revenue_np = unit_price * qty * (1 - discount)

print("First 5 revenue (NumPy):", revenue_np[:5])
print("Same as the Excel Revenue column?", np.allclose(revenue_np, df["Revenue"].to_numpy()))

# Aggregations
print("\nTotal revenue  : Rs", revenue_np.sum().round(2))
print("Average        : Rs", revenue_np.mean().round(2))
print("Median         : Rs", np.median(revenue_np))
print("Std deviation  : Rs", revenue_np.std().round(2))
print("Max / Min      :", revenue_np.max(), "/", revenue_np.min())

i = revenue_np.argmax()                       # position of the maximum
print("\nBiggest order is", order_ids[i], "with Rs", revenue_np[i])
````

````python
# Find orders with 3 or more items

bulk_orders = 0
bulk_revenue = 0

for i in range(len(qty)):

    if qty[i] >= 3:
        bulk_orders = bulk_orders + 1
        bulk_revenue = bulk_revenue + revenue_np[i]

print("Orders with 3+ items:", bulk_orders)
print("Revenue from bulk orders: Rs", round(bulk_revenue, 2))


# Label each order as Bulk or Regular

order_type = []

for q in qty:

    if q >= 3:
        order_type.append("Bulk")
    else:
        order_type.append("Regular")

print("Order types:", order_type)


# Find the highest 5 revenues

top5_revenue = sorted(revenue_np, reverse=True)[:5]

print("Top 5 revenues:", top5_revenue)
````

````python
# Create a 2D NumPy array
table = df[["Unit_Price", "Quantity", "Discount_Pct", "Revenue"]].to_numpy()

# Check the size of the array
print("Rows and columns:", table.shape)

# Show the first 3 rows
print("First 3 rows:")
print(table[:3])

# Show Revenue from the first 5 orders
print("Revenue of first 5 orders:")
print(table[:5, 3])

# Average of each column
print("Average of each column:")
print(table.mean(axis=0))
````

# Section 7: Indexing and Selection (8 min)

How do we pick the rows and columns we need?

| Method | Meaning | Example |
|---|---|---|
| `df["col"]` | one column | `df["Revenue"]` |
| `df[["a","b"]]` | many columns | `df[["City","Revenue"]]` |
| `.iloc[ ]` | by **position** (numbers) | `df.iloc[0:3]` |
| `.loc[ ]` | by **label** (names) | `df.loc[0, "City"]` |
| Boolean filter | by **condition** | `df[df["Quantity"] > 2]` |

### 📖 Definitions
| Term | Definition | Starbucks example |
|---|---|---|
| **Indexing** | Picking a specific element, row or column using its label or position. | `df["Revenue"]` |
| **Selection** | Choosing a subset of data (rows and/or columns). | Only Pune rows |
| **`.loc[ ]`** | Selects by **label** (names). End of slice **is** included. | `df.loc[0:2, "City"]` |
| **`.iloc[ ]`** | Selects by **integer position**. End of slice is **not** included. | `df.iloc[0:3]` |
| **Boolean filtering** | Selecting rows where a condition is `True`. | `df[df["Revenue"] > 800]` |
| **`&` , `\|`** | AND / OR to combine conditions. Each condition must be inside brackets. | Mumbai **and** Cold Coffee |
| **`.isin()`** | Checks if values belong to a given list. | UPI or Wallet |
| **`.sort_values()` / `.nlargest()`** | Sort rows / get the top N rows. | Top 5 orders |
| **`.query()`** | Filter using a text condition. | `df.query("City == 'Pune'")` |
| **`set_index()`** | Makes one or more columns the row labels. | OrderID as index |
| **MultiIndex** | An index with more than one level. | (City, Product) |

# Section 8: Basic Data Structures for Organising Business Datasets (10 min)

Real companies keep data in **several tables**, not one big table. Our Excel file follows this design:

| Table (Excel sheet) | Type | Contains |
|---|---|---|
| `Products` | Master table | Product, Category, Base_Price, **Cost_Price** |
| `Stores` | Master table | StoreID, Store_Name, City, **Region**, **Store_Type** |
| `Sales_Data` | Transaction table | One row per order |

We join them with `merge` (like VLOOKUP) and summarise with `groupby` and `pivot_table`.

### 📖 Definitions
| Term | Definition | Starbucks example |
|---|---|---|
| **Dataset** | A collection of related data, usually in table form. | The Excel file |
| **Master table** | A table that stores stable reference information. | `Products`, `Stores` sheets |
| **Transaction table** | A table that records business events (sales, orders). | `Sales_Data` sheet |
| **Key / ID column** | A common column used to link two tables. | `StoreID`, `Product` |
| **Merge (Join)** | Combining two tables using a common key column. Similar to VLOOKUP in Excel. | Add Cost_Price and Region to sales |
| **GroupBy** | Splitting data into groups and applying a calculation to each group. | Revenue for each city |
| **Aggregation (`agg`)** | Summarising each group with sum, mean, count, etc. | Orders, revenue, avg rating per region |
| **Pivot table** | A summary table where one column becomes rows and another becomes columns. Same as Excel Pivot Table. | City × Category revenue |
| **Margins** | Row and column totals added to a pivot table. | TOTAL row |
| **Revenue** | Money earned from sales = Price × Quantity × (1 - Discount). | Rs 300 × 2 × 0.8 = Rs 480 |
| **Profit** | Revenue minus cost. | Revenue - (Cost_Price × Quantity) |
| **Profit margin** | Profit as a percentage of revenue. | Profit / Revenue × 100 |

````python
print("PRODUCTS (master):")
print(products_df)
print("\nSTORES (master):")
print(stores_df)
````

````python
# MERGE: bring Cost_Price (from Products) and Region/Store details (from Stores) into the sales table
full = (df.merge(products_df[["Product", "Cost_Price"]], on="Product", how="left")
          .merge(stores_df[["StoreID", "Store_Name", "Region", "Store_Type"]], on="StoreID", how="left"))

print("Rows before:", len(df), "| Rows after:", len(full), "(must stay equal)")
print("Columns before:", df.shape[1], "| Columns after:", full.shape[1])

# New business measures
full["Total_Cost"] = full["Cost_Price"] * full["Quantity"]
full["Profit"]     = full["Revenue"] - full["Total_Cost"]

print(full[["OrderID", "Store_Name", "Region", "Product", "Revenue", "Total_Cost", "Profit"]].head())
````

````python
# GROUPBY: split into groups, then calculate for each group
print("Revenue and profit by CITY:")
by_city = full.groupby("City")[["Revenue", "Profit"]].sum().round(2).sort_values("Revenue", ascending=False)
print(by_city)

print("\nRevenue by PRODUCT CATEGORY:")
by_category = full.groupby("Category")["Revenue"].sum().round(2).sort_values(ascending=False)
print(by_category)
````

````python
# Many measures at once with agg()
by_region = full.groupby("Region").agg(
    Orders        = ("OrderID", "count"),
    Total_Revenue = ("Revenue", "sum"),
    Avg_Order     = ("Revenue", "mean"),
    Total_Profit  = ("Profit", "sum"),
    Avg_Rating    = ("Rating", "mean"),
).round(2)
print(by_region)

print("\nTop 5 stores by revenue:")
print(full.groupby("Store_Name")["Revenue"].sum().round(2).sort_values(ascending=False).head(5))
````

````python
# Monthly trend (order the months correctly)
by_month = (full.groupby("Month")["Revenue"].sum().round(2)
                .reindex(["July", "August", "September"]))
print("Revenue by month:")
print(by_month)

print("\nPayment mode share (%):")
print((full["Payment_Mode"].value_counts(normalize=True) * 100).round(1))

print("\nAverage order value by customer type:")
print(full.groupby("Customer_Type")["Revenue"].mean().round(2))
````

````python
# PIVOT TABLE: City (rows) x Category (columns), like an Excel Pivot Table
pivot = full.pivot_table(values="Revenue", index="City", columns="Category",
                         aggfunc="sum", fill_value=0, margins=True, margins_name="TOTAL").round(0)
print(pivot)

print("\nOrders count: Order_Channel x Payment_Mode")
print(full.pivot_table(values="OrderID", index="Order_Channel", columns="Payment_Mode",
                       aggfunc="count", fill_value=0))
````

````python
# Business insights in one go
margin = (full.groupby("Category")["Profit"].sum() / full.groupby("Category")["Revenue"].sum() * 100).round(1)
print("Profit margin % by category:")
print(margin.sort_values(ascending=False))

best_store = full.groupby("Store_Name")["Profit"].sum().idxmax()
best_prod  = full.groupby("Product")["Profit"].sum().idxmax()
low_rating = full.groupby("Product")["Rating"].mean().idxmin()

print("\nMost profitable store   :", best_store)
print("Most profitable product :", best_prod)
print("Lowest-rated product    :", low_rating)
````

````python
# Save the summary tables into a NEW Excel file (3 sheets)
with pd.ExcelWriter("Starbucks_Summary.xlsx") as writer:
    by_city.to_excel(writer, sheet_name="By_City")
    by_region.to_excel(writer, sheet_name="By_Region")
    pivot.to_excel(writer, sheet_name="City_x_Category")

print("Saved: Starbucks_Summary.xlsx")
````
